```python
text = "机器学习让机器学会学习"

# 1. 统计每个字出现了几次
counts = {}

# 在这里写你的循环
for word in text:

    if word in counts:
        counts[word]+=1
    else:
        counts[word]=1


print(counts)

# 2. 找出出现次数最多的字和它的次数
most_common_char = None
most_common_count = None

# 在这里写你的代码
max_count = 0
for char in counts:
    if counts[char] > max_count:
        max_count = counts[char]
        most_common_char = char

print(most_common_char, max_count)

```

```python
# 第 1 题：返回列表的平均值；如果列表为空，返回 0
def get_average(numbers):
    if numbers==[]:
        return 0
    result=sum(numbers)/len(numbers)
    return result


# 第 2 题：统计文字中每个字出现的次数，并返回字典
def count_chars(text):
    count={}
    for word in text:
        if word in count:
            count[word]+=1
        else:
            count[word]=1
    return count


# 第 3 题：传入“姓名: 分数”的字典，返回分数最高的学生姓名
def get_top_student(scores):
    top_number=0
    top_name=None
    for name, number in scores.items():
        if number>top_number:
            top_number = number
            top_name = name
        
    return top_name


print(get_average([80, 90, 100]))
print(get_average([]))
print(count_chars("猫狗猫"))
print(get_top_student({"小明": 88, "小红": 95, "小刚": 76}))
```
