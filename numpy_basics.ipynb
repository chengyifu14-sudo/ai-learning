```python
import numpy as np

names = np.array(["小明", "小红", "小刚", "小丽", "小强"])
scores = np.array([
    [88, 92, 79],
    [95, 67, 90],
    [58, 73, 61],
    [85, 85, 91],
    [45, 58, 62],
])
avg=scores.mean(axis=1).round(2)
print(avg)
pos=avg.argmax()
print(names[pos], avg.max())
print(scores.max(axis=0))
print(scores[scores>=90])
print(len(scores[scores>=90]))
imp=scores[:,2]-scores[:,0]
print(imp)
w = np.array([0.3, 0.3, 0.4])
total=scores@w
print(total)
```
