```python
import numpy as np

X = np.array([
    [90, 85, 70],
    [80, 78, 75],
    [60, 62, 85],
    [70, 65, 80],
    [95, 92, 65],
    [50, 55, 90],
])   # 6 行（学生）× 3 列（数学、物理、语文）
mean=X.mean(axis=0)
Xc=X-mean
print("平均分",mean.round(2))
cov=Xc.T@ Xc /(len(X)-1)
print("协方差矩阵：\n",cov.round(1))
vals,vecs=np.linalg.eigh(cov)
order=np.argsort(vals)[::-1]
vals=vals[order]
vecs=vecs[:,order]
print("特征值:", vals.round(2))
print("每个方向占的比例:", (vals / vals.sum()).round(3))
print("第一主成分方向:", vecs[:, 0].round(3))
Z = Xc @ vecs[:, :1]
print("压缩后:\n", Z.round(2))
```
