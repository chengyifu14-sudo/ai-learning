```python
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
result={}
for i in range (len(words)):
    if words[i] in result:
        result[words[i]]+=1
    else:
        result[words[i]]=1
print(result)

# 应该返回 {'apple': 3, 'banana': 2, 'cherry': 1}
```

```python
students = {
    "小明": [80, 75, 90],
    "小红": [55, 88, 92],
    "小刚": [60, 61, 70],
    "小李": [90, 45, 30],
}
passed=[]
for names, scores in students.items():
    all_pass = True
    for score in scores:
        if score<60:
            all_pass=False
    if all_pass==True:
        passed.append(names)

print(passed)

# 应该打印 ['小明', '小刚']
```

标记变量模板

什么时候用：要判断一组东西是不是“全部都满足”某个条件（全部及格、全部是正数、全部不重复……）。
结果 = []
for 名字, 一组数据 in 字典.items():
    all_ok = True                  # 1. 先假设全部满足（每组重新假设）
    for x in 一组数据:
        if x 不满足条件:
            all_ok = False         # 2. 找到一个反例就推翻
    if all_ok:                     # 3. 全部看完后再下结论
        结果.append(名字)

```python
students = {
    "小明": [80, 75, 90],
    "小红": [55, 88, 92],
    "小刚": [60, 61, 70],
    "小李": [90, 45, 30],
}
top_score=0
top_name=None
ave=[]
for names, scores in students.items():
    ave=sum(scores)/len(scores)
    if ave>top_score:
        top_score=ave
        top_name=names
print(top_name)
print(round(top_score,2))
# 应该打印 小明 81.67
```
