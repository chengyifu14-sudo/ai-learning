```python
class Wallet:
    def __init__(self, owner, money):
        self.owner = owner
        self.money = money

    def add(self, amount):
        self.money += amount

    def show(self):
        return f"{self.owner} 有 {self.money} 元"

w1 = Wallet("小明", 100)
w2 = Wallet("小红", 50)
w1.add(30)
print(w1.show())
print(w2.show())
```

```python
class Student:
    # 造学生时自动运行：把姓名和成绩存到学生身上
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    # 算平均分，并交出结果
    def average(self):
        return round(sum(self.scores) / len(self.scores), 2)

    # 往这个学生的成绩列表里加一门成绩
    def add_score(self, score):
        self.scores.append(score)

    # 判断是否每门都及格
    def all_passed(self):
        all_pass = True
        for s in self.scores:
            if s < 60:
                all_pass = False
        return all_pass


s = Student("小红", [95, 67, 90])
print(s.name)
print(s.average())
print(s.all_passed())
s.add_score(58)
print(s.scores)
print(s.average())
print(s.all_passed())
```

```python
class Cart:
    def __init__(self,owner,prices):
        self.owner=owner
        self.prices=prices
    def total(self):
        return sum(self.prices)
    def add_item(self,price):
        self.prices.append(price)
    def count_over(self, boundary):
        count=0
        for num in self.prices:
            if boundary<=num:
                count+=1
        return count

c = Cart("小明", [30, 120, 75])
print(c.owner)
print(c.total())
print(c.count_over(50))
c.add_item(200)
print(c.prices)
print(c.total())
print(c.count_over(100))
```
