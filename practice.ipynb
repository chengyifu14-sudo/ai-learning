```python
def count_chars(text):
    counts={}
    for letter in text:
        if letter not in counts:
            counts[letter]=1
        else:
            counts[letter] += 1
    return counts
print(count_chars("hello"))
```

```python
students = {
    "小明": [88, 92, 79],
    "小红": [95, 67, 90],
    "小刚": [58, 73, 61],
    "小丽": [85, 85, 91],
    "小强": [45, 58, 62],
}
passed=[]
for name, scores in students.items():
    all_pass = True
    for score in scores:
        if score<60:
            all_pass =False
    if all_pass is True: 
        passed.append(name)

print(passed)

            
```
